```python
%%capture_text --path "etc_ttest_pvalues.txt"
telelist = ['NAO', 'NPGO', 'PDO', 'PNA', 'ENSO','AO']

for telecon in telelist:
    n = '       '
    pt2m_list = ''
    pmsl_list = ''
    ptp_list = ''
    pd2m_list= ''
    pwv_list = ''
    for month in [None,10,11,12,1,2,3]:
        
        
        if month is None:
            p_mask = table[f'{telecon} Phase'] == 'Positive'
            n_mask = table[f'{telecon} Phase'] == 'Negative'
        else:
            p_mask = (table[f'{telecon} Phase']=='Positive')&(table['month_strongest_GLR']==month)
            n_mask = (table[f'{telecon} Phase']=='Negative')&(table['month_strongest_GLR']==month)

        # Use the mask to filter rows
        pos = table[p_mask]
        neg = table[n_mask]


        t2m_p = pos['t2m']
        t2m_n = neg['t2m']
        t2m_diff= np.nanmean(t2m_p)-np.nanmean(t2m_n)

        msl_p = pos['msl']
        msl_n = neg['msl']
        msl_diff= np.nanmean(msl_p)-np.nanmean(msl_n)

        tp_p = pos['tp']
        tp_n = neg['tp']
        tp_diff= np.nanmean(tp_p)-np.nanmean(tp_n)
        
        d2m_p = pos['d2m']
        d2m_n = neg['d2m']
        d2m_diff= np.nanmean(d2m_p)-np.nanmean(d2m_n)

        tcwv_p = pos['tcwv']
        tcwv_n = neg['tcwv']
        tcwv_diff= np.nanmean(tcwv_p)-np.nanmean(tcwv_n)
        
        n = n+str(len(tcwv_p))+','+str(len(tcwv_n))+'|'

        t2mstat, t2m  = stats.ttest_ind(t2m_p, t2m_n)
        mslstat, msl  = stats.ttest_ind(msl_p, msl_n)
        tpstat, tp    = stats.mannwhitneyu(tp_p, tp_n)
        d2mstat, d2m  = stats.ttest_ind(d2m_p, d2m_n)
        tcwvstat,tcwv = stats.ttest_ind(tcwv_p, tcwv_n)
        
        string=[]
        for var in [t2m,msl,tp,d2m,tcwv]:
            if var < 0.055:
                string.append('\033[1m')
                string.append('\033[0m')
            else:
                string.append('')
                string.append('')
        
        pt2m_list = pt2m_list+' '+string[0]+f'{t2m_diff:0.2f}({t2m:0.3f})'+string[1]
        pmsl_list = pmsl_list+' '+string[2]+f'{msl_diff/100:0.2f}({msl:0.3f})'+string[3]
        ptp_list = ptp_list+' '+string[4]+f'{tp_diff*1000:0.2f}({tp:0.3f})'+string[5]
        pd2m_list = pd2m_list+' '+string[6]+f'{d2m_diff:0.2f}({d2m:0.3f})'+string[7]
        pwv_list = pwv_list+' '+string[8]+f'{tcwv_diff:0.2f}({tcwv:0.3f})'+string[9]
    
    print('Telecon:','\033[1m'+telecon+'\033[0m','  OCT ','  NOV','  DEC','  JAN','  FEB','  MAR')
    print(n)
    print('t2m:  ', pt2m_list)
    print('msl:  ', pmsl_list)
    print('tp:   ',  ptp_list)
    print('d2m:  ', pd2m_list)
    print('tcwv: ', pwv_list)
    print('')
    
```
